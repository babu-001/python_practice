```python
def flatten_list(lst):
    result=[]
    for i in lst:
        if isinstance(i,list):
            result.extend(flatten_list(i))
        else:
            result.append(i)
    return result
nested_list=[1,2,[3,[4,5,6]],8,9]
print(flatten_list(nested_list))
```

```python
def count_vowels_consonant(s):
    vowels="aeiouAEIOU"
    v_count=0
    c_count=0
    for i in s:
        if i.isalpha():
            if i in vowels:
                v_count+=1
            else:
                c_count+=1
    return v_count,c_count
print(count_vowels_consonant("asdfghj"))
```

```python
def is_sorted(lst):
    for i in range(len(lst)-1):
        if lst[i]>lst[i+1]:
            return False
        return True
lst=list(map(int,input().split()))
print(is_sorted(lst))
```
